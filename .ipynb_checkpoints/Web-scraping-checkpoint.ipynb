```python
import pandas as pd
import os
import time
import random
import requests
from bs4 import BeautifulSoup
import unicodedata  # Keep in import section

def strip_accents(text):
    """Remove accents/diacritics (e.g., Dončić -> Doncic) for safer matching."""
    return ''.join(c for c in unicodedata.normalize('NFD', text)
                   if unicodedata.category(c) != 'Mn')

def normalize_name(text):
    """Lowercase + strip + remove accents for robust player-name matching."""
    return strip_accents(text).strip().lower()

# -----------------------
# Player list & seasons
# -----------------------
players = [
    # Superstars / MVP-caliber
    "LeBron James", "Kevin Durant", "Stephen Curry", "Nikola Jokić",
    "Giannis Antetokounmpo", "Joel Embiid", "Kawhi Leonard", "James Harden",
    "Russell Westbrook",

    # All-Stars / Franchise players
    "Jayson Tatum", "Luka Dončić", "Anthony Davis", "Devin Booker",
    "Damian Lillard", "Jimmy Butler", "Kyrie Irving", "Paul George", "Chris Paul",
    "Donovan Mitchell", "Trae Young", "Jamal Murray", "Khris Middleton",
    "Rudy Gobert", "Nikola Vučević", "Zach LaVine", "Jaren Jackson Jr.",
    "Evan Mobley", "OG Anunoby", "Jamal Murray",

    # Rising / Great players
    "Shai Gilgeous-Alexander", "Ja Morant", "Zion Williamson", "LaMelo Ball",
    "Anthony Edwards", "Tyrese Haliburton", "Cade Cunningham", "Jalen Green",
    "Scottie Barnes", "Victor Wembanyama", "Chet Holmgren", "Klay Thompson",
    "Bradley Beal", "DeMar DeRozan", "Mike Conley", "Julius Randle",
    "Jaylen Brown", "Jrue Holiday", "Bam Adebayo", "Andrew Wiggins",
    "Kristaps Porzingis", "Deandre Ayton", "CJ McCollum", "Michael Porter Jr.",
    "Desmond Bane", "RJ Barrett", "Austin Reaves", "Derrick White",

    # Solid starters / Key contributors
    "De'Aaron Fox", "Domantas Sabonis", "Karl-Anthony Towns", "Brandon Ingram",
    "Pascal Siakam", "Fred VanVleet", "Draymond Green", "Brook Lopez",
    "Mikal Bridges", "Aaron Gordon", "Tyrese Maxey", "Darius Garland",
    "Jamal Murray", "Jalen Brunson", "Franz Wagner", "Alperen Sengun",
    "Josh Giddey", "Bennedict Mathurin", "Keegan Murray", "Immanuel Quickley",
    "Cam Thomas", "OG Anunoby", "Zion Williamson", "Herb Jones",
    "Jarrett Allen", "De'Andre Hunter", "Buddy Hield", "Tobias Harris",
    "Kyle Kuzma", "Kristian Porzingis",  # safe alias if typo occurs elsewhere
]

# De-duplicate in case of accidental repeats above
players = list(dict.fromkeys(players))

years = list(range(2016, 2026))  # 2016–2025 seasons

# -----------------------
# Output directory
# -----------------------
os.makedirs("output", exist_ok=True)

# -----------------------
# Request headers
# -----------------------
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}

def get_table(url):
    """
    Fetch first HTML table from a page. Returns DataFrame or None.
    Uses bytes -> UTF-8 decode to avoid encoding issues.
    """
    try:
        response = requests.get(url, headers=HEADERS, timeout=20)
        response.raise_for_status()
        html = response.content.decode('utf-8', errors='ignore')
        return pd.read_html(html)[0]
    except Exception as e:
        print(f"⚠️ Error fetching {url}: {e}")
        return None

def pick_team_row(df_rows):
    """
    If a player has multiple rows (traded mid-season), prefer the '2TM' row.
    Otherwise take the first occurrence.
    Works with either 'Tm' or 'Team' column names.
    """
    team_col = 'Tm' if 'Tm' in df_rows.columns else ('Team' if 'Team' in df_rows.columns else None)
    if team_col is None:
        return df_rows.iloc[[0]]
    if team_col in df_rows.columns and '2TM' in df_rows[team_col].astype(str).values:
        return df_rows[df_rows[team_col].astype(str) == '2TM']
    return df_rows.iloc[[0]]

def get_col(df, *names):
    """
    Access column by trying multiple candidate names (e.g., 'Tm' vs 'Team').
    Raises KeyError if none exists.
    """
    for n in names:
        if n in df.columns:
            return df[n]
    raise KeyError(f"None of the columns {names} found. Available: {list(df.columns)}")

# -----------------------
# Scrape loop
# -----------------------
all_data = []

for year in years:
    print(f"Processing season {year-1}-{str(year)[-2:]} ...")
    # Random polite delay between seasons
    time.sleep(random.uniform(3, 6))

    # 1) Per-game
    url_pg = f"https://www.basketball-reference.com/leagues/NBA_{year}_per_game.html"
    df_pg = get_table(url_pg)
    if df_pg is None:
        continue
    df_pg = df_pg[df_pg["Player"] != "Player"]  # drop header rows repeated in table

    # 2) Advanced
    url_adv = f"https://www.basketball-reference.com/leagues/NBA_{year}_advanced.html"
    df_adv = get_table(url_adv)
    if df_adv is None:
        continue
    df_adv = df_adv[df_adv["Player"] != "Player"]

    for player in players:
        # Match rows for this player (normalize for accents/case/whitespace)
        pg_rows = df_pg[df_pg["Player"].apply(normalize_name) == normalize_name(player)].copy()
        if pg_rows.empty:
            continue
        pg = pick_team_row(pg_rows)

        adv_rows = df_adv[df_adv["Player"].apply(normalize_name) == normalize_name(player)].copy()
        if adv_rows.empty:
            continue
        adv = pick_team_row(adv_rows)

        # Team and position columns may vary in name ('Tm' vs 'Team')
        team_val = get_col(pg, 'Tm', 'Team').values[0]
        pos_val = get_col(pg, 'Pos').values[0]

        # Build unified row
        merged = pd.DataFrame({
            "Player": [player],
            "Season": [f"{year-1}-{str(year)[-2:]}"],
            "Team": [team_val],
            "Position": [pos_val],

            # Box score (from per-game)
            "PTS": pd.to_numeric(get_col(pg, "PTS").values[0], errors="coerce"),
            "TRB": pd.to_numeric(get_col(pg, "TRB").values[0], errors="coerce"),
            "AST": pd.to_numeric(get_col(pg, "AST").values[0], errors="coerce"),
            "STL": pd.to_numeric(get_col(pg, "STL").values[0], errors="coerce"),
            "BLK": pd.to_numeric(get_col(pg, "BLK").values[0], errors="coerce"),
            "TOV": pd.to_numeric(get_col(pg, "TOV").values[0], errors="coerce"),
            "FT":  pd.to_numeric(get_col(pg, "FT").values[0],  errors="coerce"),

            # Shooting percentages (from per-game)
            "FG%": pd.to_numeric(get_col(pg, "FG%").values[0], errors="coerce"),
            "3P%": pd.to_numeric(get_col(pg, "3P%").values[0], errors="coerce"),
            "2P%": pd.to_numeric(get_col(pg, "2P%").values[0], errors="coerce"),
            "FT%": pd.to_numeric(get_col(pg, "FT%").values[0], errors="coerce"),

            # Advanced (from advanced)
            "TS%":  pd.to_numeric(get_col(adv, "TS%").values[0], errors="coerce"),
            "BPM":  pd.to_numeric(get_col(adv, "BPM").values[0],  errors="coerce"),
            "OBPM": pd.to_numeric(get_col(adv, "OBPM").values[0], errors="coerce"),
            "DBPM": pd.to_numeric(get_col(adv, "DBPM").values[0], errors="coerce"),
            "PER":  pd.to_numeric(get_col(adv, "PER").values[0],  errors="coerce"),
            "WS":   pd.to_numeric(get_col(adv, "WS").values[0],   errors="coerce"),
        })

        all_data.append(merged)

# -----------------------
# Save output
# -----------------------
if all_data:
    df_all = pd.concat(all_data, ignore_index=True)
    df_all.to_csv("output/nba_selected_stats.csv", index=False, encoding='utf-8-sig')
    print("Saved to output/nba_selected_stats.csv")
else:
    print("No data collected.")
```
